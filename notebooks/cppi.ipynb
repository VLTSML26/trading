# Constant Proportion Portfolio Insurance (CPPI)

### 1. Limits of diversification and insurance
It is acknowledged that one of the perks of diversification is its ability di eliminate unrewarded idiosyncratic risk, that is, the capacity to minimize the risk arising from specific behaviors of the assets that form a portfolio of investment.
However, in severe bear market situations (e.g. the 2008 crysis) even well diversified portfolios can lose significant fractions of their value due to intrinsic correlations in the market.

A risk factor that simultaneously impacts every asset is called a systematic risk: the main limit of diversification is the fact that it can never reduce such risk factor.

The simplest way to avoid the downsides coming from systematic risk sources is hedging, that is, a strategy employed to offset losses in investments by taking opposite positions in related assets.
However, the reduction in risk provided by hedging also typically results in a reduction in potential profits, meaning that downside protection comes with giving up on the upsides.
While this strategy can be suitable for wealthy investors, most asset owners need access to upside performance in order to alleviate secondary costs such as taxes and commissions.

```python
sp_500 = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'META', 'JPM', 'JNJ', 'V', 'PG', 'TSLA']
ew = Portfolio(sp_500, period="max")
# msr = ew.copy().msr(rf=0.03)
```

```python
rebs = ['6M', 'Y']
cppis = [CPPI(ew, {'floor': 0.65, 'rebalance': reb, 'type': 'max dd'}) for reb in rebs]
fig, axs = plt.subplots(1, 2, figsize=(12, 5))  # Create a figure and axes array

# Flatten axs for easy iteration if more than one row/column
axs = axs.flatten()

# Iterate over each CPPI instance and plot on the corresponding axes
for ax, cppi in zip(axs, cppis):
    cppi.plot_strategy_floor_rebalance(ax=ax)  # Pass the axis to plot on

# Adjust layout
plt.tight_layout()
plt.show()
```
